```python
# convert the traasaction date in Date time Format
df["Transaction Date"] = pd.to_datetime(df["Transaction Date"],errors="coerce")
```

```python
print(df["Transaction Date"].dtype)
```

## Daily and Monthly Sales Trend:- 

```python
print("------------------------------Daily Sales-------------------------------------------------------------------------------")
daily_sales = df.groupby("Transaction Date")["Total Spent"].sum()
plt.figure(figsize=(10, 5))
daily_sales.plot()
plt.title("Daily Sales Trend")
plt.xlabel("Transaction Date")
plt.ylabel("Total Spent")
plt.show()
```

```python
print("--------------------------------Monthly Sales-----------------------------------------------------------------------------")
# first you have to add new column in Dataset i.e. Month column because we want the monthly transaction
'''
    ---.dt allows Pandas to work with date/time information.
    ---df["Transaction Date"].dt.year (get the year), df["Transaction Date"].dt.month (get the month), df["Transaction Date"].dt.day(get the day).
    ---.to_period("M") means Month, It converts each date into a year-month period. e.g. 2025-01-05  →  2025-01
    '''
df["Month"] = df["Transaction Date"].dt.to_period("M")
df

##Added new column in the Dataset
```

```python
## Monthly Sales

monthly_sales = df.groupby("Month")["Total Spent"].sum()
plt.figure(figsize=(8, 4))
monthly_sales.plot(marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Spent")
plt.xticks(rotation=45)
plt.show()
```

## 2) Correlation Matrix:-
    -- Correlation analysis is a technique used in EDA to understand the relationship between two or more numerical variables.
## Correlation value
Correlation is usually represented by a value between -1 and +1:

    +1  ----> Perfect positive relationship
    +0.7 ---> Strong positive relationship
    +0.3 ---> Weak positive relationship
    0 -------> No linear relationship
    -0.3 ----> Weak negative relationship
    -0.7 -----> Strong negative relationship
    -1 --------> Perfect negative relationship

```python
#df[["Quantity", "Price Per Unit", "Total Spent"]].corr()
corr = df.corr(numeric_only =True)
corr
```

## Correlation Heatmap: - A heatmap makes this easier to understand:

```python
plt.figure(figsize = (6,4))
sns.heatmap(df.corr(numeric_only = True),
            annot = True)
plt.title("Correlation Heatmap")
plt.show()
```

## Top-selling items
 - You can calculate the total quantity sold for each item:

```python
# 
top_items = df.groupby("Item")["Quantity"].sum().sort_values(ascending=False)
print(top_items)
```

```python
# Visualize it:
plt.figure(figsize=(9, 5))
top_items.plot(kind="bar")
plt.title("Top-Selling Items")
plt.xlabel("Item")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=45)
plt.show()
```

## Highest revenue items
 - Revenue represents the total amount of money generated from sales.
 - This analysis helps us understand the revenue generated by each item and payment method.

```python
revenue_by_item = df.groupby("Item")["Total Spent"].sum().sort_values(ascending=False)
print(revenue_by_item)
```

```python
# visualize it:- 
plt.figure(figsize=(9, 5))
revenue_by_item.plot(kind="bar")
plt.title("Revenue by Item")
plt.xlabel("Item")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```
